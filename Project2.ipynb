**Objective:**
Analyze data distributions using histograms, KDE plots, and boxplots. Compare groups (Region A vs Region B), identify outliers, evaluate skewness and spread, and summarize the findings.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

np.random.seed(42)
```

```python
# Generate sample data
n = 500

regions = np.random.choice(['Region A', 'Region B'], size=n)

values = []

for region in regions:
    if region == 'Region A':
        value = np.random.normal(70, 10)
    else:
        value = np.random.normal(80, 15)

    values.append(round(value, 2))

df = pd.DataFrame({
    "Region": regions,
    "Score": values
})

df.head()
```

```python
df.to_csv("project2_data.csv", index=False)

print("Dataset saved successfully!")
```

```python
# Histogram of Scores

plt.figure(figsize=(8,5))

plt.hist(df["Score"], bins=20, edgecolor="black")

plt.title("Histogram of Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig("histogram_scores.png", dpi=300)

plt.show()
```

```python
# KDE Plot

x = np.linspace(df["Score"].min(), df["Score"].max(), 200)

kde = gaussian_kde(df["Score"])

plt.figure(figsize=(8,5))

plt.plot(x, kde(x), linewidth=2)

plt.title("KDE of Scores")
plt.xlabel("Score")
plt.ylabel("Density")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig("kde_scores.png", dpi=300)

plt.show()
```

```python
# Box Plot by Region

region_a = df[df["Region"] == "Region A"]["Score"]
region_b = df[df["Region"] == "Region B"]["Score"]

plt.figure(figsize=(8,5))

plt.boxplot(
    [region_a, region_b],
    labels=["Region A", "Region B"]
)

plt.title("Score Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Score")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig("boxplot_regions.png", dpi=300)

plt.show()
```

```python
# Statistical Analysis

# Skewness
skewness = df["Score"].skew()

# Quartiles
Q1 = df["Score"].quantile(0.25)
Q3 = df["Score"].quantile(0.75)
IQR = Q3 - Q1

# Outlier limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df["Score"] < lower_limit) | (df["Score"] > upper_limit)]

print("Skewness:", round(skewness, 2))
print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Lower Limit:", round(lower_limit, 2))
print("Upper Limit:", round(upper_limit, 2))
print("Number of Outliers:", len(outliers))
```

```python
# Compare average scores by region

comparison = df.groupby("Region")["Score"].agg(["count", "mean", "median", "std"])

print(comparison)
```
