```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Make the random data reproducible
np.random.seed(42)
```

```python
# Generate dates for 2 years
dates = pd.date_range(start='2023-01-01', end='2024-12-31', freq='D')

# Product categories
categories = ['Electronics', 'Groceries', 'Fashion', 'Home & Kitchen']

# Empty list to store data
data = []

# Generate realistic sales
for date in dates:
    for category in categories:

        # Base sales for each category
        if category == 'Electronics':
            base = 800
        elif category == 'Groceries':
            base = 1500
        elif category == 'Fashion':
            base = 700
        else:
            base = 500

        # Weekend sales boost
        weekend_boost = 1.2 if date.weekday() >= 5 else 1

        # Random daily variation
        sales = base * weekend_boost + np.random.randint(-100, 100)

        data.append([date, category, sales])

# Create DataFrame
df = pd.DataFrame(data, columns=['Date', 'Category', 'Sales'])

# Display first 5 rows
df.head()
```

```python
# Save the dataset as a CSV file
df.to_csv("sales_data.csv", index=False)

print("Dataset saved successfully as sales_data.csv")
```

```python
# Convert Date column to datetime (if not already)
df["Date"] = pd.to_datetime(df["Date"])

# Group sales by month
monthly_sales = df.groupby(pd.Grouper(key="Date", freq="ME"))["Sales"].sum()

# Plot the line chart
plt.figure(figsize=(12,6))

plt.plot(monthly_sales.index,
         monthly_sales.values,
         marker='o',
         linewidth=2)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)

plt.xticks(rotation=45)

plt.tight_layout()

# Save the chart
plt.savefig("monthly_sales.png", dpi=300)

# Display the chart
plt.show()
```

```python
# Group sales by quarter
quarterly_sales = df.groupby(pd.Grouper(key="Date", freq="QE"))["Sales"].sum()

# Create the line chart
plt.figure(figsize=(10,5))

plt.plot(
    quarterly_sales.index,
    quarterly_sales.values,
    marker='o',
    linewidth=2,
    color='green'
)

plt.title("Quarterly Sales Trend", fontsize=16)
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.grid(True)

plt.xticks(rotation=45)

plt.tight_layout()

# Save chart
plt.savefig("quarterly_sales.png", dpi=300)

# Show chart
plt.show()
```

```python
# Calculate total sales for each category
category_sales = df.groupby("Category")["Sales"].sum()

# Create the bar chart
plt.figure(figsize=(8,5))

plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category", fontsize=16)
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=15)

plt.tight_layout()

# Save the chart
plt.savefig("category_bar.png", dpi=300)

# Display the chart
plt.show()
```

```python
# Total sales by category
category_sales = df.groupby("Category")["Sales"].sum()

# Create the pie chart
plt.figure(figsize=(7,7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Market Share by Category")

plt.tight_layout()

# Save the chart
plt.savefig("category_pie.png", dpi=300)

# Display the chart
plt.show()
```
